## Example notebook showing how to read an Icechunk store into Xarray
Note: You will need RO credentials to access the carbonplan-srm bucket.


### Imports

```python
import icechunk
import xarray as xr
from srm import catalog
```

### View datasets in the srm catalog

```python
catalog
```

### Get dataset location information

```python
cesm_g6 = catalog.get("CESM-G6-1.5K-icechunk")
print(f"uri:    {cesm_g6.uri}")
print(f"bucket: {cesm_g6.bucket}")
print(f"prefix: {cesm_g6.prefix}")
```

## Load CESM2-WACCM-SSP245 Icechunk store into Xarray

```python
storage = icechunk.s3_storage(
    bucket=cesm_g6.bucket,
    prefix=cesm_g6.prefix,
    from_env=True,
)
repo = icechunk.Repository.open(storage)
session = repo.readonly_session("main")
ds = xr.open_zarr(session.store)
print(ds)
```

```python
# Open the Icechunk store with Xarray
ds = xr.open_zarr(session.store)
print(ds)
```
